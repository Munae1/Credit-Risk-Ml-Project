The missing values of 3116 and 895 together is about 12.5% of my dataset, so leaving this data out may affect my results. I will instead use the median for both columns and add a flag column to tell the real values from the filled-in ones. 

```python
df_clean = df.copy() 
#Create a copy of the original dataframe to clean and preprocess
```

```python
df_clean = df_clean.drop_duplicates(keep='first')
#drop duplicate rows, keeping the first occurrence
```

```python
df_clean = df_clean[df_clean["person_age"] <= 100]
#Remove rows where age is greater than 100 (impossible ages of 123 and 144 years old)
```

```python
# Keep rows where employment length is 60 or below, OR where it is missing
# Each condition needs its own brackets () so they combine into one filter
# The whole condition sits inside ONE pair of square brackets []
df_clean = df_clean[(df_clean["person_emp_length"] <= 60) | (df_clean["person_emp_length"].isna())]
```

```python
#Add flag columns to indicate missing values in the "person_emp_length" and "loan_int_rate" columns
df_clean["emp_length_missing"] = df_clean["person_emp_length"].isna().astype(int)
df_clean["int_rate_missing"] = df_clean["loan_int_rate"].isna().astype(int)
```

```python
df_clean["person_emp_length"] = df_clean["person_emp_length"].fillna(df_clean["person_emp_length"].median())
df_clean["loan_int_rate"] = df_clean["loan_int_rate"].fillna(df_clean["loan_int_rate"].median())
# Fill missing values in the "person_emp_length" and "loan_int_rate" columns with the median of each column
```

```python
df_clean = df_clean.reset_index(drop=True) #Reset the index of the cleaned dataframe
```

These last 6 lines of code have been used to create a working copy of the raw data ("df_clean") adjusted for the following:
- Removed the 165 duplicate rows
- Removed the 5 rows of impossible ages that have been deemed as data entry errors 
- Removed the 2 rows which impossible employment lengths 
- Flagged and filled missing values with median values 
- Reset the row numbers after removing rows
